# Sampler Comparisons with Fixed and Variable Dimensions

This notebook runs two controlled radial-velocity comparisons. The fixed-dimensional experiment fits a simulated 40-epoch sinusoid with the same likelihood and uniform priors for PCAT, emcee (Foreman-Mackey et al. 2013), and dynesty (Speagle 2020), then compares marginal posteriors, mode recovery, effective sample size, likelihood evaluations, runtime, and dynesty's evidence. The variable-dimensional experiment fits a simulated 60-epoch, two-planet system with zero to three planets allowed. PCAT samples all planet counts in one run; dynesty requires a separate evidence calculation per count. emcee is excluded from this comparison because it is fixed-dimensional. The results demonstrate different capabilities and computational costs rather than a universal sampler ranking.
